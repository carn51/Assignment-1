```python
outliers_dict = {}

#per column then apply to every column
def outliers(df, col):
  up_thresh = np.percentile(df[col], 75) + (stats.iqr(df[col])* 1.5)
  low_thresh = np.percentile(df[col], 25) - (stats.iqr(df[col]) * 1.5)
  #dynamically name lists
  list_name = f"outliers_list_{col}"
  #initilise empty ones
  outliers_dict[list_name] = []

  for val in df[col]:
    if (val < low_thresh) or (val > up_thresh):
      outliers_dict[list_name].append(val)
  #print(outliers_dict)

for cols in df.select_dtypes(include = ['number']).columns:
  outliers(df, cols)
```

```python
all_outliers = []
for col_list in outliers_dict.values():
  #ai assisted
  #adds to end of tuple
  all_outliers.extend(col_list)

sorted_list = sorted(list(set(all_outliers)))
#proportion
print(len(sorted_list) / len(df.index))
```

```python
df = df[~df['Participant ID'].index.isin(sorted_list)]
df.head()
```

```python
skewness_dict = {}

#per column then apply to every column
def skewness(df, col):
  thresh = 3 *(np.mean(df[cols]) - np.median(df[cols])) / np.std(df[cols])
  #dynamically name lists
  list_name = f"skewness_list_{col}"
  #initilise empty ones
  skewness_dict[list_name] = []

  for val in df[col]:
    if val > thresh:
      skewness_dict[list_name].append(val)
  #print(outliers_dict)

for cols in df.select_dtypes(include = ['number']).columns:
  skewness(df, cols)
```

```python
all_skewness = []
for col_list in skewness_dict.values():
  #ai assisted
  #adds to end of tuple
  all_skewness.extend(col_list)

sorted_list_skew = sorted(list(set(all_skewness)))
#proportion
print(len(sorted_list_skew))
#keep
```
